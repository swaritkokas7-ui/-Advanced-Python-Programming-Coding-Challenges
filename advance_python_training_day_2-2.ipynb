```python
# Inside the abstract class we can declare the constructor
# Abstract class constructor is executed but object is not created

from abc import ABC, abstractmethod

class Test(ABC):
    def __init__(self):
        print("Constructor is called...")

    @abstractmethod
    def m1(self):
        pass


class MyPython(Test):
    def m1(self):
        print("m1 Method implementation")


m = MyPython()
m.m1()
```

```python
class Area:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def returnArea(self):
        return self.length * self.breadth


length = int(input("Enter length: "))
breadth = int(input("Enter breadth: "))

obj = Area(length, breadth)
print("Area:", obj.returnArea())
```

```python
arr = [7, 3, 9, 2, 8]

largest = second_largest = float('-inf')

for num in arr:
    if num > largest:
        second_largest = largest
        largest = num
    elif num > second_largest and num != largest:
        second_largest = num

print("Second Largest:", second_largest)
```

```python
arr = [2, 0, 1, 2, 1, 0]

low = 0
mid = 0
high = len(arr) - 1

while mid <= high:

    if arr[mid] == 0:
        arr[low], arr[mid] = arr[mid], arr[low]
        low += 1
        mid += 1

    elif arr[mid] == 1:
        mid += 1

    else:
        arr[mid], arr[high] = arr[high], arr[mid]
        high -= 1

print("Sorted Array:", arr)
```

```python
# We can use seek() function to move cursor (file pointer) to
# specified location.

data = "Learning python is very easy from Ashish Sir"

f = open("myfile.txt", "w")
f.write(data)

with open("myfile.txt", "r+") as f:
    filedata = f.read()
    print(filedata)

    print("Current Cursor Position is =", f.tell())

    f.seek(15)

    print("Current Cursor Position is =", f.tell())

    f.write("-JAVA-")

    f.seek(0)

    filedata = f.read()
    print(filedata)
```

```python
import os,sys
fname = input("Enter file name: ")
if os.path.isfile(fname):
    print("File exists:", fname)
    f = open(fname, 'r')
else:
    print("File does not exist:", fname)
    sys.exit(0)

lcount =wcount=ccount = 0

for line in f:
    lcount += 1
    ccount += len(line)
    words = line.split()
    wcount += len(words)
print("Number of lines:", lcount)
print("Number of words:", wcount)
print("Number of characters:", ccount)
```
