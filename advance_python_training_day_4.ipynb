```python
n = int(input())
arr = list(map(int, input().split()))

# Use the actual length of the list to prevent index errors
actual_length = len(arr)
result = []

for i in range(actual_length):
    next_greater = -1

    for j in range(i + 1, actual_length):
        if arr[j] > arr[i]:
            next_greater = arr[j]
            break

    result.append(next_greater)

print(*result)
```
