```python

num = 123
ones = num % 10
tens = (num // 10) % 10
hundreds = num // 100
reversed_num_math = (ones * 100) + (tens * 10) + hundreds
print("Original:", num)
print("Reversed (Math):", reversed_num_math)
reversed_num_slice = int(str(num)[::-1])
print("Reversed (Slicing):", reversed_num_slice)
```

```python
class Student:
    def __init__(self):
        self.rollno = int(input("Enter roll number: "))
        self.name = input("Enter name: ")

        self.m1 = int(input("Enter paper 1 marks: "))
        self.m2 = int(input("Enter paper 2 marks: "))
        self.m3 = int(input("Enter paper 3 marks: "))
        self.m4 = int(input("Enter paper 4 marks: "))
        self.m5 = int(input("Enter paper 5 marks: "))

    def calculate(self):
        total = self.m1 + self.m2 + self.m3 + self.m4 + self.m5
        percentage = total / 5

        print("Roll No:", self.rollno)
        print("Name:", self.name)
        print("Total:", total)
        print("Percentage:", percentage)

        if percentage >= 85:
            print("Student can take admission")
        else:
            print("Student can't take admission")


s1 = Student()
s1.calculate()
```
